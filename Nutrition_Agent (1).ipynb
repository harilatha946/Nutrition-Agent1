```python
import json

def nutrition_agent(user_prompt):

    system_prompt = """
    You are NutriCart, an intelligent AI Nutrition Shopping Assistant.

    Your responsibilities:

    1. Answer nutrition-related questions.
    2. Explain calories, protein, carbohydrates, fat and sugar.
    3. Help users compare grocery products.
    4. Recommend healthier grocery alternatives.
    5. Suggest high-protein and low-sugar products.
    6. Explain product ingredients.
    7. Help users make informed grocery shopping decisions.

    Guidelines:

    - Understand the user's prompt before answering.
    - Use simple English.
    - Give structured answers.
    - Never invent product nutrition facts.
    - If exact product information is unavailable,
      clearly mention that.
    - Distinguish verified product data from general
      nutritional knowledge.
    - Do not make medical claims.
    """

    response = client.chat.completions.create(

        model="openai/gpt-oss-20b",

        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": user_prompt
            }
        ],

        temperature=0.3
    )

    return response.choices[0].message.content
```

```python
models = client.models.list()

for model in models.data:
    print(model.id)
```

```python
user_prompt = """
Suggest 5 healthy snacks.

Requirements:
- High protein
- Low sugar
- Affordable
- Suitable for everyday consumption

Include product names and nutritional information.
"""

result = nutrition_agent(user_prompt)

print(result)
```

```python
import gradio as gr

def chat_with_nutrition_agent(prompt):

    if not prompt.strip():
        return "Please enter your nutrition question."

    answer = nutrition_agent(prompt)

    return answer


app = gr.Interface(
    fn=chat_with_nutrition_agent,

    inputs=gr.Textbox(
        lines=4,
        label="Ask NutriCart",
        placeholder="Example: Suggest high protein snacks under ₹100"
    ),

    outputs=gr.Markdown(label="Nutrition Assistant"),

    title="🥗 NutriCart AI",
    description="Your AI-powered nutrition and grocery shopping assistant."
)

app.launch(share=True)
```
