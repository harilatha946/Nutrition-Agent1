```python
import json

def nutrition_agent(user_prompt):

    global conversation_history
    global conversation_summary

    system_prompt = """
    Acts as a Professional
    """

    conversation_history = conversation_history + "\nUser: " + user_prompt

    messages = [
        {
            "role": "system",
            "content": system_prompt +
            "\nPrevious conversation summary: " + conversation_summary +
            "\nPrevious conversation: " + conversation_history
        },
        {
            "role": "user",
            "content": user_prompt
        }
    ]

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=messages,
        temperature=0.3
    )

    answer = response.choices[0].message.content

    conversation_history = conversation_history + "\nAssistant: " + answer

    return answer
```
